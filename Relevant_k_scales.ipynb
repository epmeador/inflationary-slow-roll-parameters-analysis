## NEQs=8 Truncation Examination

This will be a document that will explore models within the NEQs=8 truncation closure condition. We will be investigating the range of $\Delta$$N$ used here. This will be the first set ran with only certain k values, as opposed to the full spectrum. The k-value scale we are interested in is on that goes from CMB to SPACE:LISA/DECIGO to PTA scales.

To determine the correst e-folding amongst other values use:

```python
import numpy as np
import pandas as pd

c = 2.99792458e8          # m/s
Mpc_m = 3.085677581e22    # m
h = 0.7

# Rough present-day wavelength for the N=0/end-of-inflation mode
lambda0_m = 1.24e-2       # m

def freq_to_scales(f_hz, h=0.7, lambda0_m=1.24e-2):
    """
    Input:
        f_hz: frequency in Hz

    Returns:
        wavelength in m,
        k in Mpc^-1,
        k in h Mpc^-1,
        rough efold N using lambda_today = exp(N) * lambda0_m
    """
    lam_m = c / f_hz
    k_m_inv = 2 * np.pi / lam_m
    k_Mpc_inv = k_m_inv * Mpc_m
    k_hMpc_inv = k_Mpc_inv / h

    N_rough = np.log(lam_m / lambda0_m)

    return lam_m, k_Mpc_inv, k_hMpc_inv, N_rough
```

```python
k_cmb_hMpc = 0.05
k_cmb_Mpc = h * k_cmb_hMpc

f_cmb = k_cmb_Mpc / 6.467102e14

freqs = {
    "CMB pivot, k=0.05 h/Mpc": f_cmb,
    "PTA low, 1e-9 Hz": 1e-9,
    "PTA high, 1e-8 Hz": 1e-8,
    "LISA low, 1e-4 Hz": 1e-4,
    "LISA high, 1 Hz": 1.0,
    "LIGO low, 10 Hz": 10.0,
    "LIGO high, 1e4 Hz": 1e4,
}

rows = []
for name, f in freqs.items():
    lam, k_Mpc, k_hMpc, N = freq_to_scales(f, h=h, lambda0_m=lambda0_m)
    rows.append([name, f, lam, k_Mpc, k_hMpc, N])

df = pd.DataFrame(
    rows,
    columns=["Scale", "f [Hz]", "lambda [m]", "k [Mpc^-1]", "k [h Mpc^-1]", "N rough"]
)

print(df.to_string(index=False))

```
